```python
notas_alunos = ["Fernando", "Luan", "Lucas", "Pedro", "Anna Julia", "Farias", "Caio"]
notas_alunos = [5,6,4,10,9,8,10]

# Convertendo essas notas de 0 até 10 para 0 até 4
# print(notas_alunos/10 * 4) - Python puro não funciona
# print(otas_notas/10.0 * 4.0) - Python puro não funciona

for nota in notas_alunos:
    print(nota / 10 * 4)

# Mostrar somente as notas pares;
for nota in notas_alunos:
    if nota % 2 == 0:
        print("par",nota)
```

```python
# Como Criar um array Numpy? array

array_notas = np.array([7,8,9,10])

# Operações vetorizados - ADEUS E WHILE!
print(array_notas + 10)
print((array_notas * 10 / 3 + 5) * 2)
```

```python
print(array_notas)
print(array_notas > 8)
# Filtragem -> Máscara de Valores
print(array_notas[array_notas > 8])
```
