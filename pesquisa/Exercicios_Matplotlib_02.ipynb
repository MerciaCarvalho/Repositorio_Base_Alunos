### Exercício 1: Gráfico de Temperaturas Mensais

Objetivo: Praticar gráfico de linha com múltiplas séries

Crie um gráfico de linha que mostre a temperatura média de duas cidades diferentes ao longo do ano.

```python
import matplotlib.pyplot as plt

# Dados fornecidos
meses = ['Jan', 'Fev', 'Mar', 'Abr', 'Mai', 'Jun', 'Jul', 'Ago', 'Set', 'Out', 'Nov', 'Dez']
temperatura_cidade_a = [28, 27, 26, 24, 21, 19, 18, 19, 21, 23, 25, 27]
temperatura_cidade_b = [32, 31, 29, 26, 23, 21, 20, 21, 23, 26, 28, 30]

# SEU CÓDIGO AQUI:
# 1. Crie um gráfico de linha com duas séries (uma para cada cidade)
# 2. Use cores diferentes e legenda
# 3. Adicione título e rótulos nos eixos
# 4. Adicione grade para melhor leitura


plt.plot(meses, temperatura_cidade_a, marker='o', color='#BFB788', label='Cidade A')

plt.plot(meses, temperatura_cidade_b, marker='s', color='#768C4D', label='Cidade B')

plt.title('Temperatura Mensal: Cidade A, Cidade B', fontsize=14)
plt.xlabel('Meses', fontsize=12)
plt.ylabel('Temperatura (°C)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()

plt.show()
```

### Exercício 2: Popularidade de Linguagens de Programação

**Objetivo**: Praticar gráfico de barras horizontais

Crie um gráfico de barras horizontais mostrando a popularidade de diferentes linguagens de programação.

```python
# Dados fornecidos
linguagens = ['Python', 'JavaScript', 'Java', 'C#', 'PHP', 'C++']
popularidade = [45, 40, 35, 25, 20, 18]  # em porcentagem

# SEU CÓDIGO AQUI:
# 1. Crie um gráfico de barras HORIZONTAIS
# 2. Use uma paleta de cores gradiente (mais popular = cor mais forte)
# 3. Adicione os valores no final de cada barra
# 4. Inverta a ordem para mostrar a mais popular no topo

plt.barh(linguagens, popularidade, color='#B1D2D9')

# Adicionando título e rótulos
plt.title('Popularidade das Linguagens de Programação')
plt.xlabel('Linguagens')
plt.ylabel('Popularidade (%)')

for i, valor in enumerate(popularidade):
    plt.text(i, valor + 1, str(valor))

plt.barh(linguagens, popularidade, color='#B1D2D9')

plt.title('Popularidade das Linguagens de Programação')
plt.xlabel('Linguagens')
plt.ylabel('Popularidade (%)')

for i, valor in enumerate(popularidade):
    plt.text(i, valor + 1, str(valor), ha='center')

plt.show()
```

```python
# Dados do gráfico
categorias = ['A', 'B', 'C', 'D']
valores = [25, 35, 20, 20]

# Cores personalizadas (opcional)
cores = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']

# "Explodir" um pedaço (destacar)
explode = (0.05, 0.05, 0, 0)  # deslocamento de cada fatia

# Criar gráfico de pizza
plt.figure(figsize=(6, 6))  # tamanho da figura
plt.pie(
    valores,
    labels=categorias,
    colors=cores,
    explode=explode,
    autopct='%1.1f%%',  # formato das porcentagens
    startangle=90,      # começa no topo
    shadow=True         # sombra para efeito 3D
)

# Título
plt.title('Exemplo de Gráfico de Pizza', fontsize=14)

# Garantir que seja um círculo perfeito
plt.axis('equal')

# Exibir
plt.show()
```
